```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Ensure dependencies are downloaded
nltk.download('punkt')
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

def preprocess_user_message(text):
    # 1. Normalize case
    lowercased_text = text.lower()
    
    # 2. Tokenize text into structural units
    raw_tokens = word_tokenize(lowercased_text)
    
    # 3. Clean and normalize (e.g., Lemmatize and filter symbols)
    cleaned_tokens = [
        lemmatizer.lemmatize(token) 
        for token in raw_tokens 
        if token.isalnum() # removes standalone punctuation like '!' or '.'
    ]
    
    return cleaned_tokens

# Simulating a live chatbot engine
user_query = "Hey there! I am looking for your pricing plans."
processed_tokens = preprocess_user_message(user_query)

print(f"Original: {user_query}")
print(f"Tokens for Bot Logic: {processed_tokens}")
# Output Tokens for Bot Logic: ['hey', 'there', 'i', 'am', 'looking', 'for', 'your', 'pricing', 'plan']
```
