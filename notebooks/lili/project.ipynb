```python
#%pip install -U spacy en-core-web-sm
import spacy

try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    from spacy.cli import download
    download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

#intents and responses
Responses = {
    "greeting": "Hello! How can I assist you today?",
    "farewell": "Goodbye! Have a great day!",
    "thanks": "You're welcome! If you have any more questions, feel free to ask.",
    "movie_recommendation": "I can recommend some great movies! What genre are you interested in?",
    "Genre": "I can recommend movies in various genres like action, comedy, drama, horror, and more. Which one do you prefer?",
    "default": "I'm sorry, I didn't understand that. Can you please rephrase?"
}

#intents and keywords
Keywords = {
    "greeting": ["hello", "hi", "hey"],
    "farewell": ["bye", "goodbye", "see you"],
    "thanks": ["thanks", "thank you"],
    "movie_recommendation": ["recommend", "suggest", "movie"],
    "Genre": ["action", "comedy", "drama", "horror", "sci-fi", "romance", "thriller", "anime", "documentary"]

}   

def get_intent(user_input):
    doc = nlp(user_input.lower())
    for token in doc:
        for intent, keywords in Keywords.items():
            if token.text in keywords:
                return intent
    
    return "default"

def process_input(text):
    intent = get_intent(text)
    response = Responses.get(intent, Responses["default"])
    return response

while True:
    user_input = input("You: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Bot: Goodbye!")
        break
    response = process_input(user_input)
    print(f"Bot: {response}")
```
